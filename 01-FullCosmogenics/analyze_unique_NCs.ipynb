```python
"""
Check unique (evtid, nC_track_id) pairs in neutron capture simulation output.
"""
import h5py
from pathlib import Path
from typing import Set, Tuple
import numpy as np
```

```python
def get_unique_pairs(hdf5_path: Path) -> Set[Tuple[int, int]]:
    """
    Extract unique (evtid, nC_track_id) pairs from a single HDF5 file.
    
    Args:
        hdf5_path: Path to HDF5 file
        
    Returns:
        Set of unique (evtid, nC_track_id) tuples
    """
    pairs = set()
    
    with h5py.File(hdf5_path, 'r') as f:
        # Navigate to MyNeutronCaptureOutput group
        evtid_data = f['hit/MyNeutronCaptureOutput/evtid/pages'][:]
        nc_track_id_data = f['hit/MyNeutronCaptureOutput/nC_track_id/pages'][:]
        
        # Create pairs
        for evtid, nc_track_id in zip(evtid_data, nc_track_id_data):
            pairs.add((int(evtid), int(nc_track_id)))
    
    return pairs

"""Main execution function."""
# Hardcoded data directory
data_dir = Path("/home/tbueger/code/sim/RMGApplications/01-FullCosmogenics/build")  # Adjust this path as needed

# Find all matching files
hdf5_files = sorted(data_dir.glob("output_t*.hdf5"))

if not hdf5_files:
    print(f"No files matching 'output_t*.hdf5' found in {data_dir}")
    
else:
    print(f"Found {len(hdf5_files)} files")

    # Collect all unique pairs across all files
    all_pairs: Set[Tuple[int, int]] = set()

    for file_path in hdf5_files:
        try:
            pairs = get_unique_pairs(file_path)
            all_pairs.update(pairs)
            print(f"Processed {file_path.name}: {len(pairs)} unique pairs")
        except Exception as e:
            print(f"Error processing {file_path.name}: {e}")

    print(f"\nTotal unique (evtid, nC_track_id) pairs: {len(all_pairs)}")
```
